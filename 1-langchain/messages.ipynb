Messages

Messages are the fundamental unit of context for models in LangChain. They represent the input and output of models, carrying both the content and metadata needed to represent the state of a conversation when interacting with an LLM. Messages are objects that contain:

- Role - Identifies the message type (e.g. system, user)
- Content - Represents the actual content of the message (like text, images, audio, documents, etc.)
- Metadata - Optional fields such as response information, message IDs, and token usage

LangChain provides a standard message type that works across all model providers, ensuring consistent behavior regardless of the model being called.

| **Concept**         | **One-Line Definition**                                                                                    |
| ------------------- | ---------------------------------------------------------------------------------------------------------- |
| **Messages**        | Messages are structured objects that represent the conversation between the user, the AI model, and tools. |
| **Role**            | The role identifies who sent the message (System, Human, AI, or Tool).                                     |
| **Content**         | Content is the actual information or data contained in a message.                                          |
| **Metadata**        | Metadata is additional information about a message, such as token usage, IDs, timestamps, or tool calls.   |
| **SystemMessage**   | A SystemMessage provides instructions that define the model's behavior, role, and response style.          |
| **HumanMessage**    | A HumanMessage represents the user's input to the model.                                                   |
| **AIMessage**       | An AIMessage represents the model's response and may include text, metadata, and tool calls.               |
| **ToolMessage**     | A ToolMessage contains the output of a tool execution and passes it back to the model.                     |
| **Text Prompt**     | A Text Prompt is a simple string input used for single, standalone requests.                               |
| **Message Prompt**  | A Message Prompt is a list of structured messages used for conversations, memory, and tool calling.        |
| **Message History** | Message History is the collection of previous messages that provides conversational context to the model.  |
| **Tool Call**       | A Tool Call is the model's request to execute an external tool or function.                                |
| **Tool Result**     | A Tool Result is the output returned by a tool after it has been executed.                                 |


```python
import os
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    api_key=os.getenv('GROQ_API_KEY')
)

```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage("you are a poetry expert"),
    HumanMessage("write a poem on artificial intelligence ")
]

response = llm.invoke(messages)

print(response.text)
```

```python
sys_message = SystemMessage('you are a chef')

messages = [
    sys_message,
    HumanMessage('give me the recipe of cheesecake')
]

response = llm.invoke(messages)

print(response.text)
```

```python

## Detailed info to the LLM through System message
from langchain.messages import SystemMessage, HumanMessage

system_msg = SystemMessage("""
You are a senior chefr with expertise in world cuisine.
Always provide unique recepies and explain your reasoning.
Be concise but thorough in your explanations.
""")

messages = [
    system_msg,
    HumanMessage('tell me the recipe of biryani')
]

response = llm.invoke(messages)

print(response.text)

```

```python
## message metadata

human_msg = HumanMessage(
    content="Hello!",
    name="alice",
    id="msg_123"
)
```

```python
response = llm.invoke([
    human_msg
])

response
response.text
```

```python
from langchain.messages import AIMessage, SystemMessage, HumanMessage

# Create an AI message manually (e.g., for conversation history)
ai_msg = AIMessage("I'd be happy to help you with that question!")

# Add to conversation history
messages = [
    SystemMessage("You are a helpful assistant"),
    HumanMessage("Can you help me?"),
    ai_msg,  # Insert as if it came from the model
    HumanMessage("Great! What's 2+2?")
]

response = llm.invoke(messages)
print(response.content)
```

```python
response.usage_metadata
```

```python
from langchain.messages import AIMessage
from langchain.messages import ToolMessage

# After a model makes a tool call
# (Here, we demonstrate manually creating the messages for brevity)
ai_message = AIMessage(
    content=[],
    tool_calls=[{
        "name": "get_weather",
        "args": {"location": "San Francisco"},
        "id": "call_123"
    }]
)

# Execute tool and create result message
weather_result = "Sunny, 72°F"
tool_message = ToolMessage(
    content=weather_result,
    tool_call_id="call_123"  # Must match the call ID
)

# Continue conversation
messages = [
    HumanMessage("What's the weather in San Francisco?"),
    ai_message,  # Model's tool call
    tool_message,  # Tool execution result
]
response = llm.invoke(messages)  # Model processes the result
```
